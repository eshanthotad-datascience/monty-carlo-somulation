```python
import pandas as pd
import numpy as np 
 
data = pd.read_excel("datas.xlsx")
# print(data.head())
data.drop(columns=["Date"], inplace=True)
# print(data.head())

l = np.log(data / data.shift(1))
l.dropna(axis=0, inplace=True)

l2 = l.mean() # l2 goes for mean new ( μ )funtion in the equation

meu = l2.tolist()
meu
```

```python
l3 = l.std() # this is the variable which stores the standered division ( σ ) 
std = l3.to_dict()
```

```python
std
```

```python
print(type(l))
print(type(l3))
```

```python
corr_matrix = l.corr()

print(corr_matrix) # this is the finction to see the correction of stock every single day with each other how it performes with each other 
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Stock Log-Return Correlation Matrix")
plt.show()
```
